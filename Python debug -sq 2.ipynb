```python
#2.Find the error in the following Python code:
def my_function():
    x += 5
    print(x)

x = 10
my_function()
```

```python
#error:

#global keyword is required to modify a global variable inside a function

#corrected code:
x=10
def my_function():
    global x
    x += 5
    print(x)

my_function()

```
