```python
# print 0 to 20 by using range

for i in range(20):
    print(i)
```

```python
# print range 10 to 20

for i in range(10, 20):
    print(i, end=" ")
```

```python
# Print number of items in the list by using 'len'

List2 = [10, 20, 14, 55, 43, 87, 76]

print(List2)
print("Number of item in the List2:")
print(len(List2))
```

```python
a = (20, 10, 16, 19, 25, 1, 276, 188)

print(a)

for i in a:
    if i % 2 == 0:
        print(i, "is odd")
```
